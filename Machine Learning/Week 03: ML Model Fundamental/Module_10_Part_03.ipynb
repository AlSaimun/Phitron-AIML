# Module 10 – Part 03: Regression Evaluation Metrics (R², MAE, RMSE)

## Step 1: Import Libraries

```python
#Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## Step 2: Create Dataset

```python
#Create Dataset
hours = np.array([2, 4, 6, 8, 10]).reshape(-1,1)
marks = np.array([55, 60, 70, 80, 88])
hours, marks
```

## Step 3: Fit Linear Regression

```python
#Fit Linear Regression
ln_rg = LinearRegression()
ln_rg.fit(hours, marks)
pred = ln_rg.predict(hours)
pred
```

## Step 4: Compute Metrics

```python
#Compute Metrics
mae = mean_absolute_error(marks, pred)
mse = mean_squared_error(marks, pred)
rmse = np.sqrt(mse)
r2 = r2_score(marks, pred)
```

## Step 5: Print Results

```python
#Print Results
print("MAE: ", mae)
print("MSE", mse)
print("RMSE", rmse)
print("R^2:",r2)
```

## Step 6: Visualizing the Fit

```python
#Visualizing the Fit
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, pred, color="red", label="Predicted Line",linestyle="dotted", marker='o')
plt.grid(True)
plt.legend()
plt.show()
```

## Summary
- MAE: average error
- RMSE: punishes large errors
- R²: variance explained
